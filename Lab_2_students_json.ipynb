```python
import json
with open("students.json","r") as students :
  content = json.load(students)
  print(content)
  print(type(content))
```

```python
student_list = []
for x in content :
  student_list.append(x["first_name"] + " " + x["last_name"])
print(student_list)
```

```python
import ipywidgets as widgets
from IPython.display import display
dropdown = widgets.Dropdown(options = student_list,description = 'pick')
display(dropdown)

first_box = widgets.Text(description = "first name")
last_box = widgets.Text(description = "last name")
mail_box = widgets.Text(description = "email")
course_box = widgets.Text(description = "courses")
link_box = widgets.Text(description = "link")
fav_box = widgets.Text(description = "favorite")
display(first_box,last_box,mail_box,course_box,link_box,fav_box)

def show(change):
    # the for loop + if
    for x in content:
        if x["first_name"] + " " + x["last_name"] == dropdown.value:
            current = x

    # the 5 box.value lines

    first_box.value = current["first_name"]
    last_box.value = current["last_name"]
    mail_box.value = current["email"]
    course_box.value = ", ".join(current["courses"])
    link_box.value = current["link"]
    fav_box.value = current["favorite_show"]

button = widgets.Button(description="Update")
def update(change):

  for x in content:
        if x["first_name"] + " " + x["last_name"] == dropdown.value:
            current = x
  current["favorite_show"] = fav_box.value
  with open("students.json", "w") as f:
      json.dump(content, f, indent=4)


dropdown.observe(show, names='value')
show(None)
button.on_click(update)
display(button)

```
