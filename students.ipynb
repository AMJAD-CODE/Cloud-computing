```python
# @title Students Form
import json
import ipywidgets as widgets          # 🆕 library for boxes and buttons

file_path = '/content/drive/MyDrive/cloud computing/Colab Notebooks/Lab_2/students.json'

student = "Amjad sleiman" # @param ["Amjad sleiman"] {run: "auto"}
favorite_show = "arcane" # @param {"type":"string","placeholder":"Type your favorite show","run":"auto"}

with open(file_path, encoding='utf-8') as f:
    students = json.load(f)

for s in students:
    if s["first_name"] + " " + s["last_name"] == student:

        # show the details in text boxes
        display(widgets.Text(value=s["first_name"], description="First name:"))   # 🆕
        display(widgets.Text(value=s["last_name"], description="Last name:"))
        display(widgets.Text(value=s["email"], description="Email:"))
        display(widgets.Textarea(value="\n".join(s["courses"]), description="Courses:"))  # 🆕
        display(widgets.Text(value=s["link"], description="Link:"))
        display(widgets.Text(value=s.get("favorite_show", ""), description="Saved show:"))  # 🆕 .get

        # Update button
        status = widgets.Label()                     # 🆕 small text label

        def update(b, s=s):                          # 🆕 function run on click
            s["favorite_show"] = favorite_show
            with open(file_path, 'w', encoding='utf-8') as f:
                json.dump(students, f, indent=4, ensure_ascii=False)
            status.value = "✔ Saved: " + favorite_show

        button = widgets.Button(description="Update", button_style="success")  # 🆕
        button.on_click(update)                      # 🆕 connect button → function
        display(widgets.HBox([button, status]))      # 🆕 put them in one row
```
