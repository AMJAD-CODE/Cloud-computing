```python
import ipywidgets as widgets
from IPython.display import display

w = widgets.Dropdown(
    options=['Addition', 'Multiplication', 'Subtraction', 'Division'],
    value='Addition',
    description='Task:',
)

txt1 = widgets.Text()
txt1.value = "1"
display(txt1)

def on_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        print("changed to %s" % change['new'])
        txt1.value = w.value

w.observe(on_change)
display(w)
```

```python
fav_box = widgets.Text(description='Favorite:')
update_btn = widgets.Button(description='Update')
status = widgets.Output()
```

```python
students = {}   # empty dictionary

with open('students.txt', 'r') as f:
    for line in f.read().splitlines():   # one string per line, without \n
        if line.strip() == "":           # skip empty lines (safety)
            continue
        parts = line.split(',')          # cut into fields
        full_name = parts[0] + " " + parts[1]
        students[full_name] = parts      # store: name → list of fields

print(list(students.keys()))
print(students['Amjad Suli'])
```

```python
import ipywidgets as widgets
from IPython.display import display

# one dropdown, options = student names from the file
dropdown = widgets.Dropdown(options=list(students.keys()), description='Student:')

# one text box per field
first_box   = widgets.Text(description='First:')
last_box    = widgets.Text(description='Last:')
email_box   = widgets.Text(description='Email:')
courses_box = widgets.Text(description='Courses:')
link_box    = widgets.Text(description='Link:')

def show_student(name):
    parts = students[name]
    first_box.value   = parts[0]
    last_box.value    = parts[1]
    email_box.value   = parts[2]
    courses_box.value = parts[3].replace(';', ', ')   # nicer display
    link_box.value    = parts[4]
    fav_box.value = parts[5] if len(parts) > 5 else ''

def on_change(change):
    show_student(change['new'])   # change['new'] = the newly selected name

dropdown.observe(on_change, names='value')#better way to prevent mutiple updates

display(dropdown, first_box, last_box, email_box, courses_box, link_box)
show_student(dropdown.value)   # fill the boxes for the first student right away




def on_update(b):
    name = dropdown.value
    fav = fav_box.value.replace(',', ' ')        # a comma would break the file format
    new_parts = students[name][:5] + [fav]       # 5 original fields + favorite
    new_line = ','.join(new_parts)               # list → one line of text

    with open('students.txt', 'a') as f:         # 'a' = append, keep old lines
        f.write('\n' + new_line)

    students[name] = new_parts                   # update memory too
    with status:
        status.clear_output()
        print(f"Saved: {new_line}")

update_btn.on_click(on_update)
display(fav_box, update_btn, status)
```
